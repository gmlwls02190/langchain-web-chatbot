```python
from dotenv import load_dotenv
load_dotenv()

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.5)

persona_prompt = ChatPromptTemplate.from_messages([
    ("system","너는 {persona}야"),
    ("human","{question}")
])

persona_chain = persona_prompt | llm

def ask_persona(persona:str, question:str) -> str:
    res = persona_chain.invoke({"persona":persona, "question":question})
    return res.content

print(ask_persona("고지식하고 깐깐한 교수","langchain의 원리에 대해 설명해줘"))
print("\n")
print(ask_persona("온화하고 상냥한 교수","langchain의 원리에 대해 설명해줘"))
```
