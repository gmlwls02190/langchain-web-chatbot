```python
from langchain_core.runnables import RunnableLambda

def add_one(x):
    return x+1

r_add = RunnableLambda(add_one)
print(type(r_add))
print(r_add.invoke(3))
print(r_add.batch([1,2,3]))
```

```python
from langchain_core.runnables import RunnableLambda

r_add = RunnableLambda(lambda x: x+1)
r_double = RunnableLambda(lambda x: x*2)

chain_a = r_add | r_double
chain_b = r_double | r_add

print(chain_a.invoke(3))
print(chain_b.invoke(3))
print(type(chain_a))
print(type(chain_b))
```

```python
from langchain_core.runnables import (RunnableLambda, RunnableParallel, RunnablePassthrough)

analyzer = RunnableParallel(
    original = RunnablePassthrough(),
    upper = RunnableLambda(lambda x: x.upper()),
    length = RunnableLambda(lambda x: len(x)),
)

print(analyzer.invoke("hello"))
```

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
load_dotenv()

prompt = ChatPromptTemplate.from_template("{topic}에 대해서 분석가 처럼 설명해줘")

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

parser = StrOutputParser()

chain = prompt | llm | parser

print(chain.invoke({"topic":"인공지능"}))
```

```python
results = chain.batch([{"topic":"블럭체인"}, {"topic":"클라우드"}])

# for r in results:
#     print("-",r)

for chunk in chain.stream({"topic":"양자컴퓨터"}):
    print(chunk, end="", flush=True)
```

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough, RunnableLambda

result = RunnableParallel(
    original = RunnablePassthrough(),
    result = RunnableLambda(lambda x: x.strip().upper() + "!"),
)
strip = RunnableLambda(lambda x: x.strip())
upper = RunnableLambda(lambda x: x.upper())
append = RunnableLambda(lambda x: x + "!")

chain = strip | upper | append
print(chain.invoke("  hello world  "))

print(result.invoke("  hello world  "))
```

```python
result = RunnableParallel(
    total = RunnableLambda(lambda x: sum(x)),
    average = RunnableLambda(lambda x: sum(x)/len(x)),
    max = RunnableLambda(lambda x: max(x)),
)

print(result.invoke([10,20,30]))
```

```python
summary_prompt = ChatPromptTemplate.from_template("{topic}을 한줄로 요약해줘")
keyword_prompt = ChatPromptTemplate.from_template("{topic}의 주요 키워드 3개 쉼표로 구분해서 뽑아줘")
model = ChatOpenAI(model="gpt-4o-mini")
parser = StrOutputParser()
summary_chain = summary_prompt | model | parser
keyword_chain = keyword_prompt | model | parser

result = RunnableParallel(
    summary = summary_chain,
    keyword = keyword_chain,
)

print(result.invoke({"topic":"머신러닝"}))
```
